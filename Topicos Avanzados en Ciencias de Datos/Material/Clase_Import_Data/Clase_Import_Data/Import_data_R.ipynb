# Importing data in R

Loading data into R can be quite frustrating. Almost every single type of file that you want to get into R seems to require its function, and even then, you might get lost in the functions’ arguments. In short, it can be fairly easy to mix up things from time to time, whether you are a beginner or a more advanced R user.

## Importing data to R from a CSV files


### readr() function

In this section, we will read data in r by loading a CSV file from Hotel Booking Demand. This dataset consists of booking data from a city hotel and a resort hotel. To import the CSV file, we will use the readr package’s `read_csv` function. 

readr supports the following file formats with these read_*() functions:

read_csv(): comma-separated values (CSV)

read_tsv(): tab-separated values (TSV)

read_csv2(): semicolon-separated values with , as the decimal mark

read_delim(): delimited files (CSV and TSV are important special cases)

read_fwf(): fixed-width files

read_table(): whitespace-separated files

read_log(): web log files

```r
library(readr)

data1 <- read_csv('hotel_bookings_clean.csv', show_col_types = FALSE)
head(data1)
```

```r
?read_csv()
```

`read_csv()` and `read_tsv()` are special cases of the more general `read_delim()`. They're useful for reading the most common types of flat file data, comma separated values and tab separated values, respectively. `read_csv2()` uses ⁠;⁠ for the field separator and ⁠,⁠ for the decimal point. 

```r
library(readr)

data_tab <- read_delim('hotel_bookings_clean.csv', delim = ',', show_col_types = FALSE)

data_tab
```

### read.table() function

Similar to `read_csv` you can also use the `read.table` function to load the file. Make sure you are adding delimiter “,” and header = 1. It will set the first row as column names instead of “V1”, “V2”,...

```r
data2 <- read.table('hotel_bookings_clean.csv', sep=",", header = 1)
head(data2, 5)
```

### Comparing the timing of these functions

```r
timing_result <- system.time({
    data1 <- read_csv('hotel_bookings_clean.csv', show_col_types = FALSE)
})

# Print the result and timing information
print(timing_result)
```

```r
timing_result <- system.time({
    data_tab <- read_delim('hotel_bookings_clean.csv', delim = ',', show_col_types = FALSE)
})

# Print the result and timing information
print(timing_result)
```

```r
timing_result <- system.time({
    data2 <- read.table('hotel_bookings_clean.csv', sep=",", header = 1)
})

# Print the result and timing information
print(timing_result)
```

## Importing data from Excel into R


We will use the readxl package’s `read_excel` function to read a single sheet from an Excel file. The package comes with tiddyvers, but it is not the core part of it, so we need to load the package before using the function. 

The function requires the location of the data and sheet number. We can also modify how our dataframe looks by reading the description of other parameters in read_excel documentation.

```r
library(readxl)
data4 <- read_excel("hotel_bookings_clean.xlsx", sheet = 1, col_names = TRUE)
head(data4, 5)
```
